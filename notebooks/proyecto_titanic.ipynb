---
## ***Clasificacion Grupos Familiares y Cabinas***
---

```python
print(df_junto[df_junto['Pclass'] == 3]['Cabin'].value_counts(dropna=False))
```

```python
def clasificar_familia(cantidad):
    if cantidad == 0:
        return 'Grupo 1: Sin familiares'
    elif 1 <= cantidad <= 3:
        return 'Grupo 2: Familia pequeña'
    else:
        return 'Grupo 3: Familia grande'

# Aplicamos la función a la nueva columna
df_junto['Categoria_Familiar'] = df_junto['Familiares'].apply(clasificar_familia)
```

```python
# Distribución incluyendo a las personas que NO tenían cabina (NaN)
print("\n=== DISTRIBUCIÓN INCLUYENDO PASAJEROS SIN CABINA (NaN) ===")
distribucion_completa = df_junto.groupby('Categoria_Familiar')['Cabin'].value_counts(dropna=False)
display(distribucion_completa)
```

```python
# Contar cuántas cabinas (que no sean nulas) hay en cada grupo
resultado_cabinas = df_junto.groupby('Categoria_Familiar')['Cabin'].count()

print("Distribución de Cabinas por Grupo Familiar:")
print(resultado_cabinas)
```

## ***PORCENTAJE SUPERVIVENCIA SEGÚN CLASIFICACIÓN***
---

```python
df_junto['grupo_etario'] = df_junto['Age'].apply(
    lambda x: 'No Registra' if pd.isna(x) 
    else ('Niños' if x < 10 
          else ('Adolescentes' if x < 18 
                else ('Adultos' if x < 50 else 'Mayores')))
)
print("\n=== DISTRIBUCIÓN DE GRUPOS ETARIOS ===")
distribucion_etaria = df_junto['grupo_etario'].value_counts(dropna=False)
display(distribucion_etaria)    
```

```python
# Si es nulo (NaN) significa que NO tenía cabina. Si tiene texto, SÍ tenía.
df_junto['Tiene_Cabina'] = df_junto['Cabin'].apply(lambda x: 'No' if pd.isna(x) else 'Sí')
```

```python
# Agrupamos por las tres variables predictoras y calculamos la proporción de supervivencia
proporciones = df_junto.groupby(['Sex', 'grupo_etario', 'Tiene_Cabina'])['Survived'].value_counts(normalize=True) * 100

print("\n=== PROPORCIÓN DE SUPERVIVENCIA POR SEXO, EDAD Y CABINA (%) ===")
display(proporciones.to_frame(name='Porcentaje (%)'))
```

Mirando los porcentajes de la tabla queda clarísimo que salvarse no dependió de la suerte. El género, los rangos de edad y el tener una cabina asignada cambiaban todo el panorama.

Género y botes salvavidas: La regla de mujeres y niños primero se nota de inmediato. Las mujeres (dando igual si eran adultas o adolescentes) tienen casi siempre más del 64% de supervivencia, e incluso hay varios grupos donde se salvó el 100%. Con los hombres pasó todo lo contrario. Los adultos que no tenían cabina registraron un 87.41% de muertes (solo sobrevivió el 12.58%) y en los mayores sin cabina la mortalidad llegó al 93.33%.

El factor de la cabina: Esto es lo que más salta a la vista en los datos. Tener cabina te aseguraba muchas más opciones de vivir. Se ve en las mujeres adolescentes y adultas con cabina, que alcanzaron un 100% y 96.87% de supervivencia. Pero el verdadero cambio se nota en los hombres: los adultos sin cabina se murieron en un 87.41%, pero si tenían una, la supervivencia subía al 43.07%. A los niños les pasó lo mismo; con cabina se salvaron todos (100%) y sin ella quedaron mitad y mitad (50% y 50%).

Casos raros por la edad: Hay datos llamativos en las edades extremas. Por ejemplo, las mujeres mayores sin cabina se salvaron todas según los registros de la tabla. En cambio, los niños sin cabina la pasaron muy mal por la falta de espacio; tanto niños como niñas sin cabina tienen un 50% de sobrevivientes y 50% de fallecidos. Eso demuestra que no tener cabina los dejó muy desprotegidos aunque fueran pequeños.

En resumen, la tabla deja ver que aunque se buscó dar prioridad por género y edad, el dinero y la clase social (reflejados en tener cabina) fueron el filtro definitivo para alcanzar un bote.
